1. Fixed Stack Dynamic
Example: Static List of 3 Numbers

```python
def fixed_stack_dynamic_example():
    nums = [0, 0, 0]  # Fixed size
    nums[0] = 1
    nums[1] = 2
    nums[2] = 3
    print("Fixed Stack Dynamic:", nums)

fixed_stack_dynamic_example()
```

2. Stack Dynamic
Example: List Size from User Input

```python
def stack_dynamic_example():
    size = int(input("Enter size: "))
    nums = [i for i in range(size)]  # Size decided at runtime
    print("Stack Dynamic:", nums)

stack_dynamic_example()
```

3. Fixed Heap Dynamic
Example: Class with Fixed List

```python
class FixedHeapDynamic:
    def __init__(self):
        self.nums = [1, 2, 3]  # Fixed when object is created

    def show(self):
        print("Fixed Heap Dynamic:", self.nums)

obj = FixedHeapDynamic()
obj.show()
```

4. Heap Dynamic
Example: Growing List with append()

```python
def heap_dynamic_example():
    nums = []  # Dynamic
    nums.append(10)
    nums.append(20)
    nums.append(30)
    print("Heap Dynamic:", nums)

heap_dynamic_example()
```
